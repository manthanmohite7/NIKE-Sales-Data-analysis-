 ## Sorting and Filtering using Numpy 

```python
print("Gross Sales in Ascending Order:")
print(np.sort(gross_sales))
```

```python
print("Profit in Ascending Order:")
print(np.sort(profit))
```

```python
print("Gross Sales in Descending Order: ")
print(np.sort(gross_sales)[::-1])
```

```python
print("Profit in Descending Order:")
print(np.sort(profit)[::-1])
```

```python
print("Top 10 Highest Gross Sales:")
print(np.sort(gross_sales)[-10:])
```

```python
print("Top 10 Highest Profit:")
print(np.sort(profit)[-10:])
```

```python
print("Bottom 10 Gross Sales:")
print(np.sort(gross_sales)[:10])
```

```python
print("Bottom 10 Profit:")
print(np.sort(profit)[:10])
```

```python
print("Gross Sales Greater than Average:")
print(gross_sales[gross_sales > np.mean(gross_sales)])
```

```python
print("Profit Greater than Average:")
print(profit[profit > np.mean(profit)])
```

```python
print("Quantity Greater than 10:")
print(quantity[quantity > 10])
```

```python
print("Customer Rating Greater than 4:")
print(rating[rating > 4])
```

```python
print("Net Sales Greater than 5000:")
print(net_sales[net_sales > 5000])
```

```python
print("Profit Less than 1000:")
print(profit[profit < 1000])
```

## Additional NumPy Functions Used

- Boolean Indexing (`array[array > value]`) – Filtered records based on conditions.
- Slicing (`[:10]`, `[-10:]`) – Displayed the top and bottom values.

```python
import matplotlib.pyplot as plt 
```

## Bar Chart - Total Gross Sales, Net Sales, Profit and Cost

```python
categories = ["Gross Sales", "Net Sales", "Profit", "Cost"]
values = [
    np.sum(gross_sales),
    np.sum(net_sales),
    np.sum(profit),
    np.sum(cost)
]
plt.figure(figsize=(8,5))
plt.bar(categories, values, width=0.4,color="#66BB6A")
plt.title("Nike Sales Summary",color="#D90000")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.show()
```

## Pie Chart - Gross Sales vs Cost vs Profit

```python
labels = ["Gross sales", "cost", "profit"]
values = [
    np.sum(gross_sales),
    np.sum(cost),
    np.sum(profit)
]
plt.figure(figsize=(6,6))
plt.pie(values, labels=labels, autopct="%1.1f%%", startangle=90)
plt.title("sales Distribution")
plt.show()
```
